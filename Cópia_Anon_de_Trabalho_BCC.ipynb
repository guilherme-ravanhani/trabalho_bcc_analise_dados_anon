```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
arquivos = files.upload() #Desta forma basta fazer o upload do arquivo csv obtido através do google forms, lembre-se de deixá-lo com o seguinte nome: "Trabalho_BCC_Respostas.csv".
# Por isso optamos por criar um formulário no google forms, assim, ainda que surjam novas respostas no formulário, basta fazer o download da planilha atualizada e renomeá-la. O código vai permanecer funcionando.
```

```python
import io
df = pd.read_csv(io.BytesIO(arquivos["Trabalho_BCC_Respostas.csv"]))

#--- INÍCIO DA ANONIMIZAÇÃO COM FAKER ---
from faker import Faker
#Configura o Faker para gerar dados no padrão brasileiro.
fake = Faker('pt_BR')
#1. Substitui os nomes da coluna "Nome" por nomes fictícios.
if 'Nome' in df.columns:
    df['Nome'] = [fake.name() for _ in range(len(df))]
#2. Substitui os e-mails.
coluna_email = 'Nome de usuário'
if coluna_email in df.columns:
    df[coluna_email] = [fake.email() for _ in range(len(df))]
#--- FIM DA ANONIMIZAÇÃO ---

#O display agora mostrará a tabela com nomes e e-mails gerados aleatoriamente
display(df)
```

##Relacionando dados com as perguntas

```python
#@title RELAÇÃO DAS RESPOSTAS SOBRE A ETNIA DOS PARTICIPANTES E A QUESTÃO SOBRE MEU AMBIENTE DE TRABALHO SER DIVERSO OU NÃO

import seaborn as sns #importamos o seaborn para fazer o heatmap à direita do histograma;
plt.subplots(figsize=(10, 10)) #alterando o tamanho da imagem.

#Preparando os dados para criação do gráfico, fazendo a contagem das respostas por etnia.
df_2dhist = pd.DataFrame({
    x_label: grp['Meu ambiente de trabalho/estudos é diverso em relação ao gênero e etnia dos colaboradores na minha equipe/curso.'].value_counts()
    for x_label, grp in df.groupby('Com qual etnia você se identifica?')
})
sns.heatmap(df_2dhist, cmap='viridis') #faz o heatmap ao lado do gráfico utilizando os dados das perguntas, para que sirva de legenda. "cmap='viridis' é o que define o esquema de cores usado, o padrão de cor escolhido foi o viridis"
plt.xlabel('\nCom qual etnia você se identifica?')
_ = plt.ylabel("Meu ambiente de trabalho/estudos é rico em diversidade")
```

```python
#@title Meu ambiente de trabalho/estudos é diverso em relação ao gênero e etnia dos colaboradores na minha equipe/curso. Relacionado Meu gênero dificulta/dificultou minha entrada no mercado de trabalho ou ambiente de estudos.


plt.subplots(figsize=(10, 9))
df_2dhist = pd.DataFrame({
    x_label: grp['Meu gênero dificulta/dificultou minha entrada no mercado de trabalho ou ambiente de estudos.'].value_counts()
    for x_label, grp in df.groupby('Meu ambiente de trabalho/estudos é diverso em relação ao gênero e etnia dos colaboradores na minha equipe/curso.')
})
sns.heatmap(df_2dhist, cmap='viridis')
plt.xlabel('\nMeu ambiente de trabalho/estudos é diverso em relação ao gênero e etnia dos colaboradores na minha equipe/curso.')
_ = plt.ylabel('Meu gênero dificulta/dificultou minha entrada no mercado de trabalho ou ambiente de estudos.')
```
